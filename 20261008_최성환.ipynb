예를 들어 `100`은 정수, `100.0`은 실수, `"100"`은 문자열입니다. Python의 값은 모두 객체입니다.

`temperature = 13.5`에서 `temperature`는 `13.5`라는 값을 가리킵니다. Python에서는 변수를 사용하기 전에 이름에 값을 연결해야 합니다.

### 식별자(변수 이름 등)의 규칙
- 숫자로 시작할 수 없습니다: `news_count`는 가능, `1news`는 불가능합니다.
- 영문자·숫자·밑줄을 조합할 수 있으며, Python은 한글 이름도 허용합니다.
- 공백과 하이픈은 식별자에 사용할 수 없습니다: `news count`, `news-count`는 변수 이름으로 쓸 수 없습니다.
- 대소문자를 구분합니다: `news`와 `News`는 서로 다른 이름입니다.
- `if`, `for`, `class` 같은 키워드는 변수 이름으로 사용할 수 없습니다.
- `list`, `str`, `input` 등 내장 함수·타입 이름은 문법적으로 가능하지만 다른 변수 이름으로 덮어쓰지 않는 것이 좋습니다.

### 조건 표현식
`참일 때 값 if 조건 else 거짓일 때 값` 형태입니다.
이 예제의 합격 기준은 **60점 이상**, 유효한 점수 범위는 **1~100**입니다.

```python
try:
    score = int(input("점수를 입력하세요(1~100): "))
    if 1 <= score <= 100:
        result = "합격" if score >= 60 else "불합격"
        print(f"{result}했습니다.")
    else:
        print("점수는 1~100 사이여야 합니다.")
except ValueError:
    print("점수는 정수로 입력하세요.")
```

### 과정 연결 실습 ①: 기사에 쓰는 증감률
다음 수치는 **수업용 가상 데이터**입니다.

증감률(%) = `(올해 값 - 전년도 값) / 전년도 값 × 100`

기준값이 0일 때는 증감률을 계산하지 않습니다. 비율이 10%에서 12%가 되었다면 **2%p 상승**, 상대적인 증가율은 **20%**로 서로 다릅니다.

```python
# 수업용 가상 데이터: 전년도 100건, 올해 120건
previous_count = 100
current_count = 120

difference = current_count - previous_count
print(f"전년 대비 건수 변화: {difference:+d}건")

if previous_count == 0:
    print("기준값이 0이므로 증감률을 계산할 수 없습니다.")
else:
    change_rate = difference / previous_count * 100
    print(f"전년 대비 증감률: {change_rate:+.1f}%")
```

## 웹 응답과 API 데이터
### 비교 예제: HTML 응답 읽기
기존 테스트 페이지는 HTML을 반환합니다. HTML 문자열을 `json.loads()`로 처리할 수는 없습니다.
`urllib.request`를 명시적으로 가져오고 `with`로 응답을 닫습니다. `timeout=10`으로 네트워크 작업에 제한 시간을 둡니다.
이 비교 예제의 사이트 연결이 실패하더라도 아래 날씨 API 실습은 별도로 실행할 수 있습니다.

```python
import urllib.request
from urllib.error import URLError

html_bytes = 100
html_url = "http://unico2013.dothome.co.kr/crawling/get.php"

try:
    with urllib.request.urlopen(html_url, timeout=10) as response:
        html_bytes = response.read()
    print(html_bytes)  # bytes: b'...'
    print(html_bytes.decode("utf-8"))  # str: 한글을 읽을 수 있는 문자열
except (URLError, TimeoutError, OSError, UnicodeDecodeError) as error:
    print("HTML 비교 예제 처리 실패:", error)
```

### Open-Meteo 날씨 API: bytes → str → dict
- 요청 위치: 서울 지역의 좌표(위도 37.5665, 경도 126.9780)
- 요청 항목: 기온과 상대습도
- 시간대: `Asia/Seoul`
- 출처: [Open-Meteo API 문서](https://open-meteo.com/en/docs)

**데이터 해석:** `current`는 기상 모델 기반 데이터입니다. 기상청 관측값이라고 표현하지 않습니다. 반환 위치는 모델 격자에 따라 요청 좌표와 다를 수 있습니다. 한 시점의 값만으로 기후 변화 추세를 판단하지 않습니다.

1. `response.read()`로 바이트열을 읽습니다.
2. `decode("utf-8")`로 문자열로 바꿉니다.
3. `json.loads()`로 JSON 문자열을 Python 객체로 바꿉니다. 이 API의 최상위 객체는 딕셔너리입니다.

API가 실패하면 메시지를 확인하세요. 실패한 응답을 실제 날씨처럼 요약하지 않습니다.

```python
import urllib.request
from urllib.error import URLError
import json

weather_url = (
    "https://api.open-meteo.com/v1/forecast"
    "?latitude=37.5665&longitude=126.9780"
    "&current=temperature_2m,relative_humidity_2m"
    "&timezone=Asia%2FSeoul"
)

# 다시 실행할 때 이전 응답이 남지 않도록 초기화합니다.
weather_bytes = 0
weather_text =0
data = 0

try:
    with urllib.request.urlopen(weather_url, timeout=10) as response:
        weather_bytes = response.read()
    weather_text = weather_bytes.decode("utf-8")
    parsed_data = json.loads(weather_text)
    if not isinstance(parsed_data, dict):
        raise ValueError("예상한 딕셔너리 형식의 응답이 아닙니다.")
    data = parsed_data
    print(weather_bytes)
except (URLError, TimeoutError, OSError, UnicodeDecodeError, ValueError) as error:
    data = None
    print("날씨 API 처리 실패:", error)
```

```python
if data is None :
    print("위 날씨 API 셀을 성공적으로 실행한 후 확인하세요.")
else:
    print("응답 본문:", type(weather_bytes))
    print("디코딩 결과:", type(weather_text))
    print("JSON 변환 결과:", type(data))
```

```python
if data is not None :
    print(weather_text)
else:
    print("처리된 날씨 데이터가 없습니다.")
```

### 과정 연결 실습 ②: 출처가 있는 날씨 요약
위 API 셀이 성공한 뒤 실행합니다. API가 제공한 기준 시각·시간대·단위를 사용하며, 값이 없으면 요약을 보류합니다.
생성한 문장은 **실습용 요약**입니다. 실제 보도에서는 데이터 성격과 기준 시각을 검토하고 필요한 공식 자료와 교차 확인합니다.

```python
if data is None:
    print("위 날씨 API 셀을 먼저 성공적으로 실행하세요.")
else:
    try:
        current = data["current"]
        units = data["current_units"]
        temperature = current["temperature_2m"]
        humidity = current["relative_humidity_2m"]

        if temperature is None or humidity is None:
            print("일부 값이 없어 요약을 보류합니다.")
        else:
            print(
                f"[실습용 요약] {current['time']} ({data['timezone']}) 기준, "
                f"서울 지역의 모델 기온은 {temperature}{units['temperature_2m']}, "
                f"상대습도는 {humidity}{units['relative_humidity_2m']}입니다."
            )
            print("출처: Open-Meteo / 기상 모델 기반 데이터")
            print("요청 URL:", weather_url)
    except (KeyError, TypeError) as error:
        print("예상한 응답 항목이 없어 요약을 보류합니다:", error)
```

## 주요 연산자와 대입 구문 정리

| 구분 | 기호 | 의미·주의 사항 |
| --- | --- | --- |
| 산술 | `+`, `-`, `*`, `/` | 덧셈·뺄셈·곱셈·나눗셈 |
| 산술 | `//`, `%`, `**` | 바닥 나눗셈·나머지·거듭제곱. `-3 // 2`는 `-2` |
| 비교 | `==`, `!=`, `<`, `<=`, `>`, `>=` | 값 비교 |
| 논리 | `and`, `or`, `not` | 조건 조합과 부정. `and`·`or`는 피연산자 자체를 반환할 수 있음 |
| 포함 | `in`, `not in` | 문자열·컬렉션 등에 값이 포함되는지 확인 |
| 동일성 | `is`, `is not` | 같은 객체인지 확인. 일반적인 값 비교에는 `==`, `None` 확인에는 `is None` |
| 비트 | `&`, `\|`, `^`, `~`, `<<`, `>>` | 정수의 비트 연산(보충 학습) |
| 대입 구문 | `=`, `+=`, `-=`, `*=`, `/=`, `//=`, `%=`, `**=`, `&=`, `\|=`, `^=`, `<<=`, `>>=` | 이름과 객체 연결, 연산을 포함한 대입 |
| 대입 표현식 | `:=` | 표현식 안에서 이름에 값을 대입(심화 학습) |
| 행렬 곱 | `@`, `@=` | 배열 라이브러리 등에서 사용(심화 학습) |

이 표는 참고용입니다. 모든 기호를 첫 수업에서 암기할 필요는 없습니다. 대입 구문의 `=`과 값 비교 연산자 `==`를 구분하세요.

## 오류 학습: 예상 오류를 확인하는 예제
다음은 **의도적으로 오류를 발생시키는 실습**입니다. `try/except`로 예상 오류를 출력하므로 전체 실행을 계속할 수 있습니다.
직접 오류 화면을 확인하고 싶다면 오류를 일으키는 줄만 별도 셀에서 실행하세요.

### ① 값을 연결하지 않은 이름: NameError

```python
try:
    print(v0)  # v0에 값을 대입한 적이 없습니다.
except NameError as error:
    print("예상한 NameError:", error)
```

### ② 이름과 값의 개수 불일치: ValueError

```python
try:
    v1, v2, v3 = 11, 22, 33, 44  # 이름 3개에 값 4개를 대입
except ValueError as error:
    print("예상한 ValueError:", error)

# 실패한 대입 뒤에 남은 이전 값으로 결과를 판단하지 않습니다.
v1, v2, v3, v4 = 11, 22, 33, 44
print("정상 대입:", v1, v2, v3, v4)
```

### ③ del로 이름 연결을 지운 뒤 사용: NameError
`del num`은 이름의 연결을 지웁니다. 객체의 메모리가 즉시 해제된다는 의미는 아닙니다.

```python
num = 10
del num
```

```python
try:
    print(num)
except NameError as error:
    print("예상한 NameError:", error)
```
